```python
# 4.5 Temperature converter
class Temperature:
    """A utility class for converting between temperature scales.

    All methods are static because they do not need any object data
    (no self) and do not need class data (no cls). They simply take
    a number in and give a number back.
    """

    @staticmethod
    def celsius_to_fahrenheit(celsius):
        return (9 / 5) * celsius + 32

    @staticmethod
    def fahrenheit_to_celsius(fahrenheit):
        return (5 / 9) * (fahrenheit - 32)

    @staticmethod
    def celsius_to_kelvin(celsius):
        return celsius + 273.15



celsius = 25
print("1 Celsius    :", celsius)
print("2 Fahrenheit :", Temperature.celsius_to_fahrenheit(celsius))
print("3 Kelvin     :", Temperature.celsius_to_kelvin(celsius))


print("\n===== Celsius -> Fahrenheit and Kelvin =====")
for c in [-40, 0, 25, 37, 100]:
    f = Temperature.celsius_to_fahrenheit(c)
    k = Temperature.celsius_to_kelvin(c)
    print(f"{c:>5} C = {f:>7.2f} F = {k:>7.2f} K")

print("\n===== Fahrenheit -> Celsius =====")
for f in [-40, 32, 77, 98.6, 212]:
    c = Temperature.fahrenheit_to_celsius(f)
    print(f"{f:>6} F = {c:>6.2f} C")
```

```python
class Product:
    store_name = "ITC online Store"
    product_count = 0

    def __init__(self, name, price, quantity):
        if not Product.is_valid_price(price):
            raise ValueError("Price must be positive")
        self.name = name
        self.price = price
        self.quantity = quantity
        Product.product_count += 1
    def display_product(self):
        print(f"Name :{self.name}")
        print(f"Price : {self.price}")
        print(f"Quantity : {self.quantity}")
        print(f"Total Value : ${self.total_value()}")
    def total_value(self):
        return self.price * self.quantity

    @classmethod
    def get_product_count(cls):
        return cls.product_count
    @staticmethod
    def is_valid_price(price):
        return price > 0

product = [
    Product("Laptop", 800, 5),
    Product("Mouse", 15, 20),
    Product("Keyboard", 30, 10),
]

print(f"====== {Product.store_name.upper()}=====")
inventory_value = 0
for number, product in enumerate(product, start=1):
    print(f"\nProduct {number}")
    product.display_product()
    inventory_value += product.total_value()

print(f"\nTotal Products : {Product.get_product_count()}")
print(f"Total Inventory Value: ${inventory_value}")
```
